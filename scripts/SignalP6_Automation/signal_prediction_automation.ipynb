This notebook shows the automation of signal sequence detection using SignalP 6.0


The bash script signalp6_automation_250910.sh will take the .csv file exported from the SQL database and re-organise the accession and amino acid sequence data into a .fasta file (00.<input_file>.fa).

The .fasta file will then be analysed using SignalP 6.0 using slow-sequential mode to detect signal sequence in both "eukarya" and "other" mode.

The result of "eukarya" mode will be outputted into the folder 01.signalp_eukarya, and that of "other" mode will be outputted into the folder 02.signalp_other.

The prediction_results.txt file outputted by SignalP6 will automatically be re-organised using signal_prediction_cleaning_250910.py into a .csv file.





**Please modify signalp_path, pydir, and cleanup_prog_dir in signalp6_automation_250910.sh according to your setup.**

**Usage of signalp6_automation_250910.sh:** 

sudo bash signalp6_automation_250910.sh <input_file.csv> <output_directory>

**Usage of signal_prediction_cleaning.py:**
